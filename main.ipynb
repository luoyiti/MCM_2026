```python
# Combine all Google Trends data files
import pandas as pd
import os

data_dir = '/Users/luoyiti/Project/MCM_2026/data/google_trends'
all_files = [f for f in os.listdir(data_dir) if f.endswith('_trends.csv')]

print(f'Found {len(all_files)} trend files')

dfs = []
for f in all_files:
    try:
        df = pd.read_csv(os.path.join(data_dir, f))
        dfs.append(df)
    except Exception as e:
        print(f'Error: {f}: {e}')

if dfs:
    combined = pd.concat(dfs, ignore_index=True)
    output_path = os.path.join(data_dir, 'all_celebrities_trends.csv')
    combined.to_csv(output_path, index=False)
    print(f'Saved combined file with {len(combined)} rows')
    print(f'Unique celebrities: {combined["celebrity_name"].nunique()}')
    print(f'Date range: {combined["date"].min()} to {combined["date"].max()}')
    print()
    display(combined.head(10))
```

```python
# Check which celebrities are collected and show summary
import subprocess

# Check if collection process is still running
result = subprocess.run(['ps', 'aux'], capture_output=True, text=True)
if 'robust_collect' in result.stdout or 'collect_trends' in result.stdout:
    print("⏳ Collection process is still running in background")
else:
    print("❌ Collection process is not running")

# Show collected celebrities
collected_celebrities = combined['celebrity_name'].unique()
print(f"\n✅ Collected {len(collected_celebrities)} celebrities:")
for celeb in sorted(collected_celebrities):
    season = combined[combined['celebrity_name'] == celeb]['season'].iloc[0]
    count = len(combined[combined['celebrity_name'] == celeb])
    print(f"  - {celeb} (Season {season}): {count} data points")
```

```python
# 合并Google Trends数据到原始数据集
import pandas as pd
import os

data_dir = '/Users/luoyiti/Project/MCM_2026/data'

# 读取原始数据
original_df = pd.read_csv(f'{data_dir}/2026_MCM_Problem_C_Data.csv')
print(f"原始数据: {len(original_df)} 行, {len(original_df.columns)} 列")

# 读取Google Trends汇总数据
trends_df = pd.read_csv(f'{data_dir}/serpapi/celebrity_trends_summary.csv')
print(f"Trends数据: {len(trends_df)} 行")

# 计算方差 (variance = std^2)
trends_df['trend_variance'] = trends_df['trend_std'] ** 2

# 只保留需要的列
trends_subset = trends_df[['celebrity_name', 'trend_mean', 'trend_variance']].copy()
trends_subset.columns = ['celebrity_name', 'google_trends_mean', 'google_trends_variance']

print(f"\nTrends数据样本:")
print(trends_subset.head())

# 合并数据
merged_df = original_df.merge(trends_subset, on='celebrity_name', how='left')

# 检查合并结果
matched = merged_df['google_trends_mean'].notna().sum()
print(f"\n成功匹配: {matched}/{len(original_df)} 名人")

# 检查未匹配的
unmatched = merged_df[merged_df['google_trends_mean'].isna()]['celebrity_name'].tolist()
if unmatched:
    print(f"未匹配的名人 ({len(unmatched)}): {unmatched}")

# 保存更新后的数据
merged_df.to_csv(f'{data_dir}/2026_MCM_Problem_C_Data.csv', index=False)
print(f"\n✅ 已保存到 2026_MCM_Problem_C_Data.csv")
print(f"新列数: {len(merged_df.columns)}")

# 显示新增列的统计信息
print(f"\n新增列统计:")
print(f"google_trends_mean: min={merged_df['google_trends_mean'].min():.2f}, max={merged_df['google_trends_mean'].max():.2f}, mean={merged_df['google_trends_mean'].mean():.2f}")
print(f"google_trends_variance: min={merged_df['google_trends_variance'].min():.2f}, max={merged_df['google_trends_variance'].max():.2f}, mean={merged_df['google_trends_variance'].mean():.2f}")
```
